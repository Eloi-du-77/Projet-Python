# Introduction
Chine, Etats-Unis, Japon, Royaume-Uni... Plus de 1000 médailles à eux 4 aux derniers jeux olympiques. Pourquoi ce sont systématiquement les mêmes pays dans le top 5 des jeux olympiques ? pourquoi ces pays ont de meilleurs athlètes qu'ailleurs ? Pourquoi Tuvalu n'a jamais remporté de médaille ? La question a déjà été traitée notamment dans l'article (Is the distribution of Olympic medals truly fair ? Martin Martinez PhD 2024) (lien dans le readme). Selon lui, l'éducation est un facteur important dans la réussite sportive.

Qu'en est-il des jeux paralympiques ? **Quels sont les déterminants de la réussite d'un pays aux jeux paralympiques ?**. Nous ferons le choix de nous intéresser aux pays les plus fortunés pour voir des facteurs moins évidents que le PIB par habitant (dépenses publiques en éducation, en sport, en accompagnement aux handicapés, en aménagement du territoire). Le premier objectif sera d'établir un lien entre réussite olympique et réussite paralympique via une régression linéaire simple. Puis, nous ferons une régression ridge entre réussite paralympique et caractéristiques nationales (montant de certains types de dépense publique, IDH). Enfin, si toutes les caractéristiques ont un impact significatif, nous ferons une analyse en composantes principales pour répartie ces caractéristiques en groupes. Sinon on fera une régression simple de la réussite paralympique sur chacune des rares variables significatives.

Enfin, une conclusion nous permettra d'expliquer les résultats et d'exprimer nos contraintes dans ce travail.

NB : Dans toute l'étude, on appelera le score sportif la variable qui donne, pour chaque édition des jeux, trois points par médaille d'or, deux par médaille d'argent, une par médaille de bronze le tout ramené au nombre d'athlètes de la délégation.

## Première agrégation de la base

<div style="text-align: justify;">

```python
from Toutes_les_df_olympiques.df_medailles_paralympiques import df_para as df_medailles_paralympiques
from Toutes_les_df_olympiques.df_athletes_paralympiques import df_para as df_athletes_paralympiques
from Toutes_les_df_olympiques.df_medailles_olympiques import df_medailles as df_medailles_olympiques
from Toutes_les_df_olympiques.df_athletes_olympiques import df_athletes as df_athletes_olympiques
from Toutes_les_df_de_depenses.df_depenses_publiques import df_longue as df_depenses_publiques

df_merge = df_medailles_olympiques.merge(
    df_medailles_paralympiques, 
    on=['pays', 'annee'], 
    how='outer'
)

df_merge = df_merge.merge(
    df_athletes_olympiques, 
    on=['pays', 'annee'], 
    how='outer'
)

df_merge = df_merge.merge(
    df_athletes_paralympiques, 
    on=['pays', 'annee'], 
    how='outer'
)

df_merge = df_merge.merge(
    df_depenses_publiques, 
    on=['pays', 'annee'], 
    how='outer'
)

df_merge = df_merge.merge(
    df_education, 
    on=['pays', 'annee'], 
    how='outer'
)

df_merge = df_merge.merge(
    df_idh, 
    on=['pays', 'annee'], 
    how='outer'
)

df_merge = df_merge.merge(
    df_pib_par_habitant, 
    on=['pays', 'annee'], 
    how='outer'
)

df_merge.query("pays == 'France' and annee.isin([2012,2016,2020,2024])")
```

<a id="par_habitant"></a>
### Création de variables de médailles par athlètes et de dépenses par habitant

<div style="text-align: justify;">

Sans ramener les variables au nombre d'athlètes, l'étude n'aurait que peu d'intérêt, on ramène donc les variables sportives au nombre d'athlètes présents et les variables de dépenses en valeur absolue par habitant.

</div>

Ici le nombre de médailles par athlète aux jeux paralympiques:

```python
df_merge['athletes_olympiques'] = pd.to_numeric(df_merge['athletes_olympiques'], errors='coerce')
df_merge['athletes_paralympiques'] = pd.to_numeric(df_merge['athletes_paralympiques'], errors='coerce')
df_merge["annee"] = pd.to_numeric(df_merge["annee"], errors='coerce')

df_merge['or_paralympique_par_athlete'] = df_merge['or_paralympique'] / df_merge['athletes_paralympiques']
df_merge['argent_paralympique_par_athlete'] = df_merge['argent_paralympique'] / df_merge['athletes_paralympiques']
df_merge['bronze_paralympique_par_athlete'] = df_merge['bronze_paralympique'] / df_merge['athletes_paralympiques']
df_merge['total_medailles_paralympiques_par_athlete'] = df_merge['total_medailles_paralympiques'] / df_merge['athletes_paralympiques']

df_merge.query("pays == 'France' and annee.isin([2012,2016,2020,2024])")[['pays','annee', 'or_paralympique_par_athlete',
'argent_paralympique_par_athlete','bronze_paralympique_par_athlete','total_medailles_paralympiques_par_athlete']]
```

Et ici les dépenses par habitant dans chaque secteur :

```python
df_merge['amenagement_territoire_par_habitant'] = df_merge['amenagement_territoire'] * df_merge['pib_habitant']
df_merge['loisirs_sports_par_habitant'] = df_merge['loisirs_sports'] * df_merge['pib_habitant']
df_merge['maladie_invalidite_par_habitant'] = df_merge['maladie_invalidite'] * df_merge['pib_habitant']
df_merge['education_par_habitant'] = df_merge['education'] * df_merge['pib_habitant']

df_merge.query("pays == 'France' and annee.isin([2012,2016,2020,2024])")[['pays','annee', 'amenagement_territoire_par_habitant',
'loisirs_sports_par_habitant','maladie_invalidite_par_habitant','education_par_habitant']]
```

<a id="score_olympique"></a>
### Création de score olympique et paralympique

<div style="text-align: justify;">

Pour évaluer la performance sportive, deux variables sont immédiates : le total de médailles et le classement aux Jeux.

Cependant, le classement aux Jeux valorise trop les médailles d'or (avoir une médaille d'or et zéro médailles d'argent nous place devant un pays avec zéro médailles d'or et dix d'argent) et le total de médailles pas assez.

Nous créons donc un score olympique et paralympique rapportant trois points par médaille d'or, deux par médaille d'argent, un par médaille de bronze le tout ramené au nombre d'athlètes.

Ici la table pour le score paralympique en 2016 :

</div>

```python
df_merge['score_paralympique'] = (3 * df_merge['or_paralympique_par_athlete'] + 
                                 2 * df_merge['argent_paralympique_par_athlete'] + 
                                 df_merge['bronze_paralympique_par_athlete'])

df_merge.query("annee==2016")[['pays','annee','score_paralympique','total_medailles_paralympiques_par_athlete']].sort_values(by='score_paralympique', ascending=False) \
        .head(10)
```

<div style="text-align: justify;">

Ce score pose cependant un problème, il met trop en avant les pays ayant un athlète exceptionnel mais aucun autre athlète disputant les Jeux. L'Ouganda illustre ce problème. Il faudra en prendre compte lors de la sélection des pays.

</div>

### Le problème de table avec tous les pays : la variance des variables explicatives

<div style="text-align: justify;">

Si la table df_tous_pays contient plusieurs milliers d'observations, il reste un problème important. Les variables explicatives (PIB par habitant, IDH, dépenses par habitant) sont extrêmement variables.

Regardons le coefficient de variation, nommé ici Quotient, de cette table.

</div>

```python
from statistiques_descriptives.Stat_descr import quotient

df_tous_pays=pd.read_pickle("Toutes_les_df_agregees/df_tous_pays.pkl")

df_quotient_tous_pays=quotient(df_tous_pays,df_tous_pays)

df_quotient_tous_pays
```

<div style="text-align: justify;">

Il est inadmissible d'utiliser une base dont l'écart-type de certaines variables est supérieur à leur moyenne.

On essaie donc de prendre les dix pays les plus riches (en terme de PIB par habitant) pour avoir a priori des variables un peu moins volatiles. On fait aussi attention à ne sélectionner des pays ayant amené au moins cinq athlètes en moyenne au cours des quatre précédents jeux paralympiques. Cela limite la surreprésentation de pays avec un seul athlète qui gagne plusieurs comme l'Ouganda présenté dans la partie première agrégation de la base.

On observe ensuite leur Quotient de variation défini comme l'écart-type sur la table actuelle (donc ici sur la table df_top_10) divisé par la moyenne de df_tous_pays.

</div>

```python
df_top_10=pd.read_pickle('Toutes_les_df_agregees/df_top_10.pkl')

pays = df_top_10['pays'].unique()
print(f"Les pays de cette nouvelle table sont : \n{', '.join(pays)}")

df_quotient_top_10=quotient(df_top_10,df_tous_pays)

df_quotient_top_10
```

```python
variables = df_quotient_top_10['Variable'].values
quotient_top10 = df_quotient_top_10['Quotient (%)'].values
quotient_tous_pays = df_quotient_tous_pays[df_quotient_tous_pays['Variable'].isin(variables)]['Quotient (%)'].values

#Créer le graphique
x = np.arange(len(variables))
largeur = 0.35
fig, ax = plt.subplots(figsize=(12, 6))
barres1 = ax.bar(x - largeur/2, quotient_top10, largeur, label='df_top_10', color='#3498db')
barres2 = ax.bar(x + largeur/2, quotient_tous_pays, largeur, label='df_tous_pays', color='#2ecc71')

#Legende
ax.set_xlabel('Variable', fontsize=12, fontweight='bold')
ax.set_ylabel('Quotient (%)', fontsize=12, fontweight='bold')
ax.set_title('Comparaison Quotients entre la table df_top_10 et la table df_tous_pays', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(variables, rotation=45, ha='right')
ax.legend()
ax.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

<div style="text-align: justify;">

Cette sélection présente des résultats encourageants, tous les quotients ont diminué à l'exception de moyenne éducation qui a augmenté. Cela veut dire que les variables de df_top_10 varient moins comparé à la moyenne de df_tous_pays. Néanmoins, ils restent très importants et proches de 100. <a id="reg_lin"></a>
### Régression linéaire entre scores olympiques

<div style="text-align: justify;">

Regardons d'abord si il y a un lien entre résultats olympiques et paralympiques. Voici la régression linéaire du score paralympique sur le score olympique pour les observations (pays,année) de la base df_top_12_sans_NaN.

</div>

```python
from analyse.reg_lin_simple import regression_simple

df_resultats=regression_simple(df,'score_olympique')
df_resultats
```

<div style="text-align: justify;">

On trouve donc que pour chaque point en plus au score olympique, le score paralympique augmente de 2,4 points.
Cet effet semble considérable. En effet, les statistiques descriptives ont montré que la plupart des scores paralympiques sont entre 0 et 1. Donc si la Suède, l'un des pays les moins performants de la table, augmentait son score olympique de 0,1, il passerait au dessus de la moyenne de la table.

Le test de significativité bilatéral donne une p-valeur de 0,004 (< 1%), ce qui indique une relation statistiquement très significative entre les deux variables. On peut rejeter d'absence d'effet avec un très haut niveau de confiance.

En revanche, on voit que la performance olympique est loin d'être le seul déterminant de la performance paralympique. En effet, le R² est de 21% (21% de la variance de score_paralympique est expliquée par score_olympique). Le montant est correcte mais montre bien que d'autres facteurs pourraient avoir de l'influence comme les dépenses publiques. Nous verrons en sous-partie suivante s'il est légitime de considérer les dépenses publiques comme déterminants du score paralympique.

Enfin, l'intervalle de confiance à 95% pour β₁ est [0,8 ; 3,9]. Cet intervalle étant entièrement positif, on peut affirmer avec 95% de confiance que l'effet est strictement positif. Cependant, l'intervalle est relativement large, reflétant une certaine imprécision pour estimer l'ampleur exacte de l'effet (qui pourrait varier d'un facteur de presque 5 entre les deux bornes).

Cet intervalle assez large peut avoir plusieurs origines : tout d'abord, la base de donnée contient un nombre d'observations très faible (38) ce qui rend l'erreur standard très forte et élargit l'intervalle.

L'essentiel à retenir de cette régression est que l'on est sûrs à 99% que le score olympique a un effet positif sur le score paralympique mais que d'autres paramètres ont aussi un effet important sur ce paramètre.

</div>
